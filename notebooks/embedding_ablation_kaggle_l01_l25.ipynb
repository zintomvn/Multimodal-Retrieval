# AIC embedding ablation — L01 đến L25

Notebook launcher cho thí nghiệm OpenCLIP, SigLIP2 và Qwen3-VL chỉ trên các batch L01–L25. Qwen image embeddings phải có sẵn đến hết L25; notebook chỉ kiểm tra coverage và không sinh thêm. Attach dataset `lcdngthnh/aic-2026`, bật Internet và GPU rồi chọn **Run All**.

```python
import os, subprocess
from pathlib import Path

REPO_URL = "https://github.com/zintomvn/Multimodal-Retrieval.git"
BRANCH = "experiment/embedding-ablation-l01-l25"
REPO_DIR = Path("/kaggle/working/Multimodal-Retrieval-l01-l25")

if REPO_DIR.exists():
    subprocess.run(["git", "fetch", "origin", BRANCH], cwd=REPO_DIR, check=True)
    subprocess.run(["git", "checkout", "-B", BRANCH, f"origin/{BRANCH}"], cwd=REPO_DIR, check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", "--branch", BRANCH, REPO_URL, str(REPO_DIR)], check=True)
commit = subprocess.check_output(["git", "rev-parse", "--short", "HEAD"], cwd=REPO_DIR, text=True).strip()
print("Repository commit:", commit)

os.environ["ABLATION_BATCH_MIN"] = "1"
os.environ["ABLATION_BATCH_MAX"] = "25"
os.environ["ABLATION_REPO_BRANCH"] = BRANCH
os.environ["ABLATION_REPO_DIR"] = str(REPO_DIR)
os.environ["ABLATION_DATASET_ID"] = "aic-l01-l25-ablation"
os.environ["ABLATION_DATASET_INPUT_DIR"] = "/kaggle/input/datasets/thnhlcdng/ablation"
os.environ["ABLATION_REUSE_QWEN_ONLY"] = "1"  # không sinh thêm; L01-L25 phải có sẵn trên Zilliz

base_notebook = REPO_DIR / "notebooks/embedding_ablation_kaggle.ipynb"
print("Launching:", base_notebook)
get_ipython().run_line_magic("run", str(base_notebook))
```
